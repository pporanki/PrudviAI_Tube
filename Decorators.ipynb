## **Decorators** - allow you to modify or extend the behavior of functions without changing their code.


```python
import time


def fetch_data():
  time.sleep(2)
  print("Here I am fetching data from the database")

def update_data():
  time.sleep(1)
  print("Here I am updating data in the database")
```

```python
def extended_orig_funct(origi_function):
  def orig_func_wrapper(*args):
    start_time = time.time()
    print("some logic before original fucntion exec")
    origi_function(*args)
    end_time = time.time()
    total_time = end_time - start_time
    print("some logic after original fucntion exec")
    print(f"Total taken to exec the {origi_function.__name__} is {total_time}")
  return orig_func_wrapper
```

```python
@extended_orig_funct   # delete_data = extended_orig_funct(delete_data)
def delete_data():
  time.sleep(1)
  print("Here I am deleting data from the database")
```

```python
def extended_orig_funct(origi_function):  # Decorator function; receives the original function
    def orig_func_wrapper(*args,**kwargs):  # Wrapper function; adds extra behavior around the original function
        start_time = time.time()

        print("some logic before original function exec")

        returnval = origi_function(*args,**kwargs)  # Call the original function

        end_time = time.time()
        total_time = end_time - start_time

        print("some logic after original function exec")
        print(f"Total taken to exec the {origi_function.__name__} is {total_time}")
        #return returnval
    return orig_func_wrapper  # Return the wrapper function to replace the original function
```

```python
import time

@extended_orig_funct
def fetch_data():
  time.sleep(2)
  print("Here I am fetching data from the database")
  return "some data"

@extended_orig_funct
def update_data(tablename="employee"):
  time.sleep(1)
  print("Here I am updating data in the database")

@extended_orig_funct   # delete_data = extended_orig_funct(delete_data)
def delete_data():
  time.sleep(1)
  print("Here I am deleting data from the database")

@extended_orig_funct    # connection_DB()  --> extended_orig_func(connection_DB)
def connection_DB(connection_str):
  print("Here I am connecting to the database")
```

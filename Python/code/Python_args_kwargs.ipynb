## 2. `*args` — Variable Positional Arguments

`*args` allows a function to accept a variable number of **positional arguments**.

The name `args` is a convention. The `*` is what matters.

````python
def show_args(*args):
    print(args)
    print(args[0], args[2])  # access the individual items like a tuple
    print(type(args))
````

````python
show_args(10, 20, 30)
show_args(10, 20, 30, 40, 50)
````

### How to consume the values of args

````python
def add(*args):
    total = 0

    for number in args:  # args as tuple
        total += number

    return total
````

````python
print(add(10, 20))
print(add(10, 20, 30))
print(add(10, 20, 30, 40, 50))
````


#### `*args` collects the positional arguments into a **tuple**.

# 3. Real-World Example: Flexible Notifications

````python
def send_notification(message, recipient):
    print(f"Sending '{message}' to {recipient}")

send_notification("Server is down", "Prudvi")
````

### Requirement changes

Now we need to support:

- 1 recipient
- 2 recipients
- 10 recipients
- potentially many recipients

Instead of changing the function every time, use `*args`.

````python
def send_notification(message, *recipients):
    for recipient in recipients:
        print(f"Sending '{message}' to {recipient}")
````

````python
# One recipient still works
send_notification("Server is down", "Prudvi")
````

````python
# Multiple recipients also work
send_notification( "Server is down", "Prudvi", "John", "David", "Anil")
````

## Function call with keyword arguments

````python
def greet(name, city):
    print(name, city)

# Positional arguments
greet("Prudvi", "Hyderabad")
greet("Hyderabad", "prudvi") # the order makes sense

# Keyword arguments
greet(name="Prudvi", city="Hyderabad")
greet(city="Hyderabad",   name="Prudvi") # order doesnot matter

#mixed way of calling
greet("prudvi",city = "Hyderabad")
````

# 5. `**kwargs` — Variable Keyword Arguments

`**kwargs` is used for a variable number of **keyword arguments**.


`**kwargs` collects them into a **dictionary**.

````python
def show_kwargs(**kwargs):
    print(kwargs)
    print(type(kwargs))
````

````python
show_kwargs(name="Prudvi", age=30, city="Hyderabad")
````

### Visual model

```text
name="Prudvi"     ─┐
age=40             ├──→ **kwargs → dictionary
city="Hyderabad"  ─┘
```

### How to consume the values of kwargs

````python
def create_profile(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")
````

````python
create_profile(name="Prudvi", role="AI Architect", experience=18, location="Hyderabad")
````

# 6. `*args` vs `**kwargs`

| | `*args` | `**kwargs` |
|---|---|---|
| Accepts | Positional arguments | Keyword arguments |
| Stores as | Tuple | Dictionary |
| Example | `10, 20, 30` | `name="Prudvi"` |

### Easy memory trick

`*args` → positional → **tuple**

`**kwargs` → keyword → **dictionary**

# 7. Using Both Together

````python
def demo(*args, **kwargs):
    print("ARGS:", args)
    print("KWARGS:", kwargs)

demo( 10, 20, c= 30, city="Hyderabad", name="Prudvi" )
````

Python automatically separates:

```text
Positional arguments → *args → tuple
Keyword arguments    → **kwargs → dictionary
```

# 8. Why Decorators Use `*args` and `**kwargs`

A decorator may need to wrap functions with different parameter lists.

`*args` and `**kwargs` make the wrapper flexible.

````python
def my_decorator(func):
    def wrapper(*args, **kwargs):
        print("Before function")

        result = func(*args, **kwargs)

        print("After function")
        return result

    return wrapper

@my_decorator
def calculate(a, b):
    return a + b

print(calculate(10, 20))
````

### What is happening?

```text
Caller
  ↓
wrapper(*args, **kwargs)
  ↓
func(*args, **kwargs)
  ↓
Original function
```

The wrapper does not need to know the exact signature of the original function.
